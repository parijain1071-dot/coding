```python
#guessing game

#generate a random int within 1-100
import random
jackpot = random.randint(1,100)

guess = int(input('guess any number'))
counter = 1
while (guess != jackpot):
   if(guess < jackpot):
      print('wrong! guess higher')
   else:
      print('wrong! guess lower')
   guess = int(input('guess any number'))
counter += 1
print('correct guess')
print('attempts', counter)

#IDENTATION MATTERS THE MOST!

```
